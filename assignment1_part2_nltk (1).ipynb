## Question 2 (1 point)

Percentage of tokens equal to `whale` or `Whale`. *Returns a float (e.g. 0.5 for 0.5 %).*

```python
def question_two():
    whale_count = sum(1 for w in text1 if w in ("whale", "Whale"))
    return whale_count / len(text1) * 100


print(question_two())
```

## Question 4 (1 point)

Tokens longer than 5 characters with a frequency above 150. *Returns an alphabetically sorted list of tokens.*

```python
def question_four():
   freq = nltk.FreqDist(text1)
   return sorted(w for w in freq if len(w) > 5 and freq[w] > 150)

print(question_four())
```

## Question 5 (1 point)

The longest token and its length. *Returns a `(token, length)` tuple.*

```python
def question_five():
    longest = max(set(text1), key=len)
    return (longest, len(longest))

print(question_five())
```

## Question 6 (1 point)

Alphabetic tokens (`isalpha()`) with a frequency above 2000. *Returns a list of `(frequency, word)` tuples sorted by decreasing frequency.*

```python
def question_six():
    freq = nltk.FreqDist(text1)
    return sorted(((f, w) for w, f in freq.items() if w.isalpha() and f > 2000), reverse=True)

print(question_six())
```

## Question 7 (1 point)

Average number of tokens per sentence (`sent_tokenize` for the sentences). *Returns a float.*

```python
from nltk.tokenize import sent_tokenize, word_tokenize

def question_seven():
  sentences = sent_tokenize(moby_raw)
  return sum(len(word_tokenize(s)) for s in sentences) / len(sentences)

print(question_seven())
```

## Question 8 (1 point)

The 10 most frequent lowercase alphabetic tokens after removing the English stop words. *Returns a list of 10 `(word, frequency)` tuples sorted by decreasing frequency.*

```python
import nltk
from nltk.corpus import stopwords

def question_eight():
    nltk.download("stopwords", quiet=True)
    stop = set(stopwords.words("english"))
    words = [w.lower() for w in text1 if w.isalpha()]
    return nltk.FreqDist(w for w in words if w not in stop).most_common(10)

print(question_eight())
```

## Question 9 (1 point)

The 10 most frequent Porter stems of the alphabetic tokens. *Returns a list of 10 `(stem, frequency)` tuples sorted by decreasing frequency.*

```python
from nltk.stem import PorterStemmer

def question_nine():
  stemmer = PorterStemmer()
  stems = [stemmer.stem(w) for w in text1 if w.isalpha()]
  return nltk.FreqDist(stems).most_common(10)

print(question_nine())
```
